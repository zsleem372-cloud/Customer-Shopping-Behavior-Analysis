```python
##تحليل الإنفاق حسب الاشتراك

df.groupby("Subscription Status").agg(
    Customers=("Customer ID", "count"),
    Total_Sales=("Purchase Amount (USD)", "sum"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2)
```

```python
###تأثير الخصومات على الإنفاق
df.groupby("Discount Applied").agg(
    Customers=("Customer ID", "count"),
    Total_Sales=("Purchase Amount (USD)", "sum"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2)

```

```python
###تحليل المبيعات حسب الموسم
df.groupby("Season").agg(
    Customers=("Customer ID", "count"),
    Total_Sales=("Purchase Amount (USD)", "sum"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2).sort_values("Total_Sales",ascending=False)
```

```python
###تحليل طرق الدفع
df.groupby("Payment Method").agg(
    Customers=("Customer ID", "count"),
    Total_Sales=("Purchase Amount (USD)", "sum"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2).sort_values("Total_Sales",ascending=False)
```

```python
###تحليل المبيعات حسب النوع
df.groupby("Gender").agg(
    Customers=("Customer ID", "count"),
    Total_Sales=("Purchase Amount (USD)", "sum"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2).sort_values("Total_Sales",ascending=False)
```

```python
##أعلى 10 منتجات حسب إجمالي المبيعات.

df.groupby("Item Purchased")["Purchase Amount (USD)"].sum().sort_values(ascending=False).head(10)
```

```python
###تحليل تقييمات العملاء حسب الفئة

df.groupby("Category")["Review Rating"].mean().sort_values(ascending=False).round(2)
```

```python
###تحليل تكرار الشراء

df.groupby("Frequency of Purchases").agg(
    Customers=("Customer ID", "count"),
    Avg_Spending=("Purchase Amount (USD)", "mean")
).round(2).sort_values("Customers", ascending=False)
```
